## Question 1

### Prompt
What does this function print?


````python
def f(n):
    if n == 0:
        return
    print(n)
    f(n-1)
    print(n)

f(3)
````

### Solution
The output is:

```text
3
2
1
1
2
3
```

### Why
- The function prints `n` on the way **down** the recursion.
- Then it prints `n` again on the way **back up**.
- The base case is `n == 0`, where the function stops.


## Question 2

### Difference between a recursive and an iterative solution
- A **recursive solution** solves a problem by calling itself on a smaller version of the same problem, and it must have a **base case** to stop.
- An **iterative solution** uses a loop such as `for` or `while` and updates variables until it is finished.

### When recursion can be a bad choice
Recursion can be a bad choice when:
- the recursion goes very deep and may cause a **stack overflow** or Python recursion-depth error,
- the recursive version adds extra function-call overhead,
- or the problem is simpler and clearer with a loop.


## Question 3

### Code
```python
def analyze(n):
    for i in range(n - 200):
        print(i)

    for j in range(n*n):
        for k in range(n):
            print(j, k)
```

### Solution
**Tightest Big-O bound: `O(n^3)`**

### Why
- The first loop runs about `n` times, so it is `O(n)`.
- The nested loops run `(n^2) * n = n^3` times, so they are `O(n^3)`.
- The larger term dominates, so the whole function is **`O(n^3)`**.

```python
def strange(n):
    i = n
    while i > 1:
        for j in range(n):
            print(i, j)
        i = i // 2
```

### Solution
**Tightest Big-O bound: `O(n log n)`**

### Why
- The `while` loop keeps halving `i`, so it runs about `log n` times.
- Inside it, the `for` loop runs `n` times.
- Total work: `n * log n`, so the answer is **`O(n log n)`**.

## Question 11

### Compare Python lists and linked lists

#### Access time: accessing the `i`th element
- **Linked List:** `O(n)`  
  You must walk node by node from the head until you reach position `i`.
- **Python List:** `O(1)`  
  Python lists support direct indexing because they are array-based.

#### Insertion at the beginning
- **Linked List:** `O(1)`  
  You can create a new node and point it at the old head.
- **Python List:** `O(n)`  
  Every existing element must shift one position to the right.

#### Memory usage
- **Linked List:** `O(n)` space, but with **more overhead per element**  
  Each node stores both a value and a pointer/reference to the next node.
- **Python List:** `O(n)` space, usually with **less overhead per item**  
  Items are stored in a dynamic array, though Python may keep extra capacity.


## Question 12

### Write a recursive function `count_large(lst)`
Return how many values in a list are greater than 50.


````python
def count_large(lst):
    if lst == []:
        return 0

    rest_count = count_large(lst[1:])

    if lst[0] > 50:
        return 1 + rest_count
    else:
        return rest_count

# Example
count_large([30, 60, 10, 80])
````

### Why
- The **base case** is the empty list, which contains 0 values greater than 50.
- The recursive step solves the problem for the rest of the list.
- If the first item is greater than 50, add 1.

### Time complexity
This version is typically discussed as **`O(n)`** recursive work over `n` items.


## Question 13

### (a) Complete the `LinkedList` class


````python
class LinkedList:
    def __init__(self):
        # initialize an empty list
        self.head = None

    def count_greater_10(self):
        count = 0
        current = self.head

        while current is not None:
            if current.value > 10:
                count += 1
            current = current.next

        return count
````

### Why
- An empty linked list usually has its head set to `None`.
- To count nodes greater than 10, start at the head and walk through the list one node at a time.
- Increase the counter whenever a node's value is greater than 10.

### (b) Time complexity
**`O(n)`**
Because: The function visits each node exactly once.

### Write `main()`
Goals:
1. create a `TaskQueue`
2. create two dogs (`Lucy` and `GiGi`)
3. alternate tasks between them
4. stop when no tasks remain
5. print each dog's completed tasks


````python
class TaskQueue:
    def __init__(self):
        self.tasks = [
            "chase squirrel", "dig hole", "bark at doorbell", "steal snack",
            "nap in sunbeam", "zoom around house"
        ]

    def get_task(self):
        if len(self.tasks) == 0:
            return None
        return self.tasks.pop(0)

class Dog:
    def __init__(self, name):
        self.name = name
        self.completed_tasks = []

    def do_task(self, task):
        if task is not None:
            self.completed_tasks.append(task)


def main():
    queue = TaskQueue()
    dog1 = Dog("Lucy")
    dog2= Dog("GiGi")
    turn = 0

    while  (task = queue.get_task()):   #this is while task is not None
        if turn%2==0:  #take turns based on even odd
            dog1.do_task(task)
        else:
            dog2.do_task(task)
        turn +=1  

    
    print(dog1.name, dog1.completed_tasks)
    print(dog2.name, dog2.completed_tasks)


main()
````

### Here's a brief synopsis
- `queue.get_task()` removes the next task.
- The `turn` variable alternates between the two dogs.
- The loop stops when `get_task()` returns `None`.
- At the end, print each dog's completed task list.
- You should have the algorithmic chops to do this!
